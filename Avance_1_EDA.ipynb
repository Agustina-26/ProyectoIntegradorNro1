### 1.1) Imputación.

Ahora lo que haremos será darle un valor a las celdas vacías, en este caso a la celda de 'Edad'.

```python
import numpy as np          # Voy a poner un rango de edad (0 a 100) y los que estén fuera de eso, los reemplazo por NaN.

df["edad"] = pd.to_numeric(df["edad"], errors="coerce")

df.loc[(df["edad"] < 0) | (df["edad"] > 100), "edad"] = np.nan
```

```python
df['preferencias_alimenticias'] = df['preferencias_alimenticias'].fillna('DESCONOCIDO')    

# Este comando nos permite rellenar los valores nulos de la columna 'preferencias_alimenticias' por el valor 'DESCONOCIDO'.
```

```python
df['correo_electronico'] = df['correo_electronico'].fillna('NO PROPORCIONADO')
df['telefono_contacto'] = df['telefono_contacto'].fillna('NO PROPORCIONADO')

# Usamos estos comandos para rellenar los valores nulos de las columnas 'correo_electronico' y 'telefono_contacto' por el valor 'NO PROPORCIONADO'
```

```python
df['promedio_gasto_comida'] = df['promedio_gasto_comida'].fillna(df['promedio_gasto_comida'].median())

# En este caso, lo que hago es usar un comando para que la columna de 'promedio_gasto_comida' rellene lo valores nulos con la mediana de esa misma columna.
```

```python
df['edad'].isna().sum() / len(df) * 100   # los valores nulos del dataframe representan el ~1% así que no hay problemas con borrarlos.
df = df.dropna(subset=['edad'])
porcentaje_nulos = df['edad'].isna().sum() / len(df) * 100
```

```python
df.isnull().sum().sort_values(ascending=False)  

# Vuelvo a ejecutar el comando para verificar que ahora sí, no existen valores nulos en nuestro DataFrame.
```

```python
df_sin_duplicados = df.drop_duplicates()

# Ahora lo que hacemos es eliminar los valores duplicados de nuestro DataFrame, guardándolo en la nueva variable 'df_sin_duplicados'.
```

```python
display(df.head())
display(df.info())
display(df.describe()) 
```

```python
df.loc[df['frecuencia_visita'] < 0, 'frecuencia_visita'] = np.nan   

# Este comando nos ayudará a identificar los valores negativos en la columna 'frecuencia_visitas' y lo reemplazará por 'NaN'.
```

```python
df['frecuencia_visita'] = df['frecuencia_visita'].fillna(df['frecuencia_visita'].median())

# Ahora lo que hicimos fue rellenar los valores nulos (NaN) de la columna 'frecuencia_visita' con la mediana de la misma.
```

```python
display(df.head())
display(df.info())
display(df.describe())
```

```python
# Volvimos a ejecutar el comando y podemos ver que ya solucionamos el error de la columna de 'frecuencia_visita' que tenía valores negativos.
```

## Gráficos

- Como sabemos, con gráfico es más fácil entender los datos que hemos tomado.

```python
personas_ciudad = df.groupby('ciudad_residencia').size().sort_values(ascending=False)

# Queremos saber cuántas personas residen en cada ciudad. 

fig,ax = plt.subplots()
ax.bar(personas_ciudad.index, personas_ciudad.values, color = 'teal')
plt.xticks(rotation = 45)
plt.show()
```

### Lo que podemos observar en este gráfico, es que las 2 ciudades más pobladas son Chicago y Nueva York, luego la cantidad de personas disminuye bruscamente.

```python
list_est = df.groupby('estrato_socioeconomico'). size()

print(list_est)                    # Vamos a generar un  gráfico que nos permita ver cómo es la relación entre el gasto promedio mensual y el estrato socioeconómico.
```

```python
gasto_promedio_ciudad = df.groupby('ciudad_residencia')['promedio_gasto_comida'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(gasto_promedio_ciudad.index, gasto_promedio_ciudad.values, color='lightgreen')

ax.set_title('Gasto promedio mensual en restaurantes por ciudad')
ax.set_xlabel('Ciudad')
ax.set_ylabel('Promedio de gasto')
plt.xticks(rotation=45)
plt.show()
```

### Promedio de gasto vs ciudad. 
Si bien vimos que Chicago ocupa el puesto número 1 con respecto a la cantidad de personas en el gráfico anterior, en este vemos que son quienes menos gastan mensualmente. También es llamativo cómo la ciudad menos poblada (Phoenix) ocupa un puesto dentro de todo alto en lo que respecta al gasto promedio.


```python
fig, ax = plt.subplots(figsize=(10,6))

for estrato, grupo in df.groupby('estrato_socioeconomico'):
    ax.scatter(
        grupo['frecuencia_visita'],
        grupo['promedio_gasto_comida'],
        label=estrato,
        alpha=0.3,
        s = 20
    )
ax.set_title('Relación entre frecuencia de visita y gasto promedio por comida')
ax.set_xlabel('Frecuencia de visita a restaurantes')
ax.set_ylabel('Gasto promedio por comida')
ax.legend(title='Estrato socioeconómico')
plt.show()
```


### Relación entre frecuencia de visita y gasto promedio

Se observa que, a medida que aumenta el nivel socioeconómico, los clientes tienden a visitar con mayor frecuencia los restaurantes y a realizar un mayor gasto promedio.

En particular, entre las frecuencias de 2 a 4 visitas, se evidencia una diferencia significativa en el gasto entre los clientes de nivel socioeconómico medio y aquellos de nivel alto y muy alto, siendo estos últimos quienes presentan valores considerablemente superiores.

```python
plt.figure(figsize=(12,8))
sns.scatterplot(data = df,
                x = 'ingresos_mensuales',
                y = 'promedio_gasto_comida',
                hue = 'preferencias_alimenticias',
                palette = 'Set2'
                )
plt.xlabel('Ingreso Mensual')
plt.ylabel('Gasto Promedio en Comida')

plt.show()
```

### Relación entre Ingreso mensual y Gasto promedio según preferencias alimenticias. 

Podemos observar en nuestro gráfico de dispersión, que el gasto promedio se dispara luego de los 80 dólares, lo que coincide con un ingredo mensual alto. 
Si observamos por celda, podemos notar que predominan quienen tienen un consumo vegetariano; en segundo lugar, quienes consumen carnes.

```python
tabla = pd.crosstab(df['ciudad_residencia'], df['preferencias_alimenticias'])

tabla.plot(kind='barh', figsize=(10,6))
plt.xlabel('Cantidad de personas')
plt.ylabel('Ciudad')
plt.title('Distribución de preferencias alimenticias por ciudad')

plt.show()
```

### El gráfico muestra la distribución de las preferencias alimenticias de los clientes en cada ciudad, en función de la cantidad de personas que pertenecen a cada categoría.

Se puede observar como, mayoritariamente, el consumo de carne es masivo y superior por mucho a otras preferencias como lo vegetariano o los mariscos, puede reflejar diferencias culturales o hábitos entre regiones.
